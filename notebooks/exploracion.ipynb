# Exploración de las reseñas
Primer vistazo a `data/resenas.csv` antes de usar la IA. Este cuaderno no va a la web: sirve para encontrar las primeras pistas.

```python
import pandas as pd

# Cargamos las reseñas limpias
d = pd.read_csv('../data/resenas.csv', parse_dates=['fecha'])
len(d)
```

## 1. ¿Cuánto recomiendan cada juego, según lo que han jugado?

```python
# % de reseñas positivas por juego y perfil (novato, medio, veterano)
(d.pivot_table(index='juego', columns='perfil', values='recomienda', aggfunc='mean', observed=True) * 100).round(0)
```

## 2. ¿Quién escribe más: el que recomienda o el que no?

```python
# Longitud típica (mediana) del texto en caracteres
d.groupby('recomienda')['texto'].apply(lambda t: t.str.len().median())
```

## 3. ¿Qué periodo cubre cada juego?

```python
# Fecha de la reseña más antigua y la más nueva de cada juego
d.groupby('juego')['fecha'].agg(['min', 'max'])
```
